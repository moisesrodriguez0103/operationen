```python
def menue_anzeigen():
    print("Operationen:")
    print("1. Addition")
    print("2. Substraktion")
    print("3. Multiplikation")
    print("4. Division")
    print("5. Mittelwert")
    print("6. Summe einer Liste")
    print("7. Maximum einer Liste")
    print("8. Minimum einer Liste")
    print("9. Umfang eines Kreises")
    print("10. Hypotenuse eines rechtwinkligen Dreiecks")
    print("0. Beenden")

# Funktionen für die Operationen

def addition(a, b):
    return a + b

def substraktion(a, b):
    return a - b

def multiplikation(a, b):
    return a * b

def division(a, b):
    if b == 0:
        return "Fehler: Division durch Null"
    return a / b

def mittelwert(liste):
    if len(liste) == 0:
        return "Fehler: Leere Liste"
    return sum(liste) / len(liste)

def summe_liste(liste):
    return sum(liste)

def maximum_liste(liste):
    if len(liste) == 0:
        return "Fehler: Leere Liste"
    return max(liste)

def minimum_liste(liste):
    if len(liste) == 0:
        return "Fehler: Leere Liste"
    return min(liste)    

def umfang_kreis(radius):
    import math
    return 2 * math.pi * radius

def hypotenuse(a, b):
    import math
    return math.sqrt(a**2 + b**2)

def main():
    while True:
        menue_anzeigen()
        wahl = input("Wählen Sie eine Operation (0-10): ")
        
        if wahl == '0':
            print("Programm wird beendet.")
            break
        elif wahl in ['1', '2', '3', '4']:
            try:
                a = float(input("Geben Sie die erste Zahl ein: "))
                b = float(input("Geben Sie die zweite Zahl ein: "))
            except ValueError:
                print("Ungültige Eingabe. Bitte geben Sie gültige Zahlen ein.")
                continue
            if wahl == '1':
                print("Ergebnis:", addition(a, b))
            elif wahl == '2':
                print("Ergebnis:", substraktion(a, b))
            elif wahl == '3':
                print("Ergebnis:", multiplikation(a, b))
            elif wahl == '4':
                print("Ergebnis:", division(a, b))
        elif wahl in ['5', '6', '7', '8']:
            try:
                liste = input("Geben Sie eine Liste von Zahlen ein (getrennt durch Kommas): ")
                liste = [float(x.strip()) for x in liste.split(',')]
            except ValueError:
                print("Ungültige Eingabe. Bitte geben Sie gültige Zahlen ein.")
                continue
            if wahl == '5':
                print("Mittelwert:", mittelwert(liste))
            elif wahl == '6':
                print("Summe:", summe_liste(liste))
            elif wahl == '7':
                print("Maximum:", maximum_liste(liste))
            elif wahl == '8':
                print("Minimum:", minimum_liste(liste))
        elif wahl == '9':
            try:
                radius = float(input("Geben Sie den Radius des Kreises ein: "))
            except ValueError:
                print("Ungültige Eingabe. Bitte geben Sie eine gültige Zahl ein.")
                continue
            print("Umfang des Kreises:", umfang_kreis(radius))
        elif wahl == '10':
            try: 
                a = float(input("Geben Sie die Länge der ersten Seite ein: "))
                b = float(input("Geben Sie die Länge der zweiten Seite ein: "))
                print("Hypotenuse:", hypotenuse(a, b))
            except ValueError:
                print("Ungültige Eingabe. Bitte geben Sie gültige Zahlen ein.")
                continue
        else:
            print("Ungültige Auswahl. Bitte wählen Sie eine gültige Option.") 

if __name__ == "__main__":    main()

```
